```python
# Importera bibliotek
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Läs in datasetet
housing = pd.read_csv("housing.csv")

# Visa de första raderna
housing.head()
```

```python
# Grundläggande info
# Kolumner, datatyper och saknade värden
housing.info()

# Statistik för numeriska kolumner
housing.describe()
```

```python
# Kontrollera saknade värden
missing_values = housing.isnull().sum()
print("Missing values per column:\n", missing_values)
```

```python
# Utforska priser
print(housing['median_house_value'].describe())

# Histogram för prisfördelning
plt.figure(figsize=(8,5))
sns.histplot(housing['median_house_value'], bins=30, kde=True)
plt.title("Distribution of Housing Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# Pris vs yta (sqft_living)
plt.figure(figsize=(8,5))
sns.scatterplot(data=housing, x='total_rooms', y='median_house_value', alpha=0.5)
plt.title("Price vs Total Rooms")
plt.xlabel("Total Rooms")
plt.ylabel("Median House Value")
plt.show()
```

```python
# Boxplot för antal sovrum
plt.figure(figsize=(12,6))
sns.boxplot(data=housing, x='total_bedrooms', y='median_house_value')
plt.title("Price vs Number of Bedrooms")
plt.show()
```

```python
# Korrelation mellan numeriska kolumner
numeric_housing = housing.select_dtypes(include=np.number)
plt.figure(figsize=(10,8))
sns.heatmap(numeric_housing.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix (Numerical Columns Only)")
plt.show()

```
